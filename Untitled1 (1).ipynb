```python
!apt-get update > /dev/null
!apt-get install -y tesseract-ocr poppler-utils > /dev/null
!pip install pytesseract opencv-python pillow pandas numpy pdf2image matplotlib > /dev/null

import pytesseract
import cv2
import numpy as np
from PIL import Image
import pandas as pd
import re
import json
from pdf2image import convert_from_path
import matplotlib.pyplot as plt
from google.colab import files

uploaded = files.upload()
file_name = list(uploaded.keys())[0]

def load_file(file_name):
    if file_name.lower().endswith(".pdf"):
        images = convert_from_path(file_name, poppler_path="/usr/bin")
        return images[0]
    else:
        return Image.open(file_name)

image = load_file(file_name)

plt.imshow(image)
plt.axis('off')
plt.show()

def preprocess_image(image):
    img = np.array(image)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
    blur = cv2.GaussianBlur(gray, (5,5), 0)
    _, thresh = cv2.threshold(blur, 150, 255, cv2.THRESH_BINARY)
    return thresh

processed_image = preprocess_image(image)

plt.imshow(processed_image, cmap='gray')
plt.axis('off')
plt.show()

text = pytesseract.image_to_string(processed_image)
print(text)

def extract_fields(text):
    name_match = re.search(r'Name[:\- ]+([A-Za-z ]+)', text)
    name = name_match.group(1).strip() if name_match else "Not Found"

    amount_match = re.search(r'(\d+\.\d{2})', text)
    amount = float(amount_match.group(1)) if amount_match else None

    date_match = re.search(r'(\d{2}[-/]\d{2}[-/]\d{4})', text)
    date = date_match.group(1) if date_match else "Not Found"

    id_match = re.search(r'(INV\d+)', text)
    invoice_id = id_match.group(1) if id_match else "Not Found"

    return {
        "name": name,
        "amount": amount,
        "date": date,
        "invoice_id": invoice_id
    }

data = extract_fields(text)
print(data)

def validate_data(data):
    status = "valid"
    if data["name"] == "Not Found":
        status = "invalid"
    if data["amount"] is None or data["amount"] <= 0:
        status = "invalid"
    if data["date"] == "Not Found":
        status = "invalid"
    if data["invoice_id"] == "Not Found":
        status = "invalid"
    data["status"] = status
    return data

validated_data = validate_data(data)
print(validated_data)

json_output = json.dumps(validated_data, indent=4)
print(json_output)

with open("output.json", "w") as f:
    f.write(json_output)

files.download("output.json")

df = pd.DataFrame([validated_data])
display(df)
```
